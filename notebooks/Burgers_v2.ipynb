```python
x0 = np.array([1.0*dx, 0.9])

target_fun =  lambda x0: zielfunktion(x0, x, mu_train, X_train, mu_test, X_test, rank, shape, num_iter)[0]
res = minimize(target_fun, x0, method='nelder-mead', bounds=[(0, 5*dx), (-1, 2)],
               options={'disp': True, "fatol": 1e-2, "maxiter": 25})
sigma, c = res["x"]
print(sigma, c)
```
